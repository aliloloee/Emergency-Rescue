# Subject device

Simulates the device worn by somebody trapped under rubble.

A subject device only transmits when something has gone wrong, so **transmitting
is itself the distress signal** - there is no separate alarm to raise. The first
reading creates a *possible mission* and sends it to the emergency center that
owns the nearest free agent. Later readings refresh that one alert rather than
raising a new one every few seconds.

Start the **agent** notebook first. If no agent has reported a position
recently, there is nobody to route the alert to, and it waits unrouted until
there is - a later reading will deliver it with no further action here.

Transmit until interrupted.

**Stop this cell the moment the agent reaches the subject.** The device reads a
person's vital signs, so once they are out of the rubble and it comes off them
there is nothing left for it to read - a real one would simply go quiet.

If you leave it running, the system keeps raising fresh alerts, and it is right
to: a device still sending a pulse means somebody is still wearing it and still
in trouble.

The socket is closed on the way out, so whatever is still buffered gets written
rather than dropped.


```python
lattitude, longitude = "44.487309", "11.341504"

try :
    while True :
        data_subject = {
            "lat" : lattitude,
            "lng" : longitude,
            "heartrate" : random.randint(50,70),
            "timestamp" : str(datetime.now().replace(microsecond=0))
        }
        ws_subject.send(json.dumps(data_subject))
        time.sleep(3)

except KeyboardInterrupt :
    print('stopped transmitting')

finally :
    ws_subject.close()
    print("WebSocket connection closed.")
```
